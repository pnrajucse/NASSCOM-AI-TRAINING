```python
# Task 2: Synthetic Data Generation for Practice
print("\n--- Task 2: Synthetic Data Generation for Practice ---")

import pandas as pd
import numpy as np

# For reproducible results
np.random.seed(42)

# Create DataFrame
df_sales = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', periods=100, freq='D'),
    'Region': np.random.choice(['East', 'West', 'North', 'South'], size=100),
    'Product': np.random.choice(['Laptop', 'Mouse', 'Keyboard', 'Monitor'], size=100),
    'Sales': np.random.randint(100, 1001, size=100),
    'Quantity': np.random.randint(1, 11, size=100)
})

# Introduce NaN values

# 5 random NaN values in Sales
sales_nan_idx = np.random.choice(df_sales.index, size=5, replace=False)
df_sales.loc[sales_nan_idx, 'Sales'] = np.nan

# 3 random NaN values in Quantity
quantity_nan_idx = np.random.choice(df_sales.index, size=3, replace=False)
df_sales.loc[quantity_nan_idx, 'Quantity'] = np.nan

# Display first few rows
print("\nFirst 10 rows:")
print(df_sales.head(10))

# Check missing values
print("\nMissing values per column:")
print(df_sales.isna().sum())
```

```python
# Task 3: Indexing and Slicing in DataFrames
print("\n--- Task 3: Indexing and Slicing in DataFrames ---")

# Use 'df_sales' from Task 2

# 3.1 Select only the 'Region' and 'Sales' columns
df3_1 = df_sales[['Region', 'Sales']]

print("\n3.1 Region and Sales columns:")
print(df3_1.head())


# 3.2 Select rows where Product is 'Laptop'
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

print("\n3.2 Rows where Product = Laptop:")
print(df3_2.head())


# 3.3 Select rows where Region is 'East' AND Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

print("\n3.3 Rows where Region = East and Sales > 500:")
print(df3_3.head())


# 3.4 Select the Sales value for the first entry where Product is 'Monitor'

monitor_rows = df_sales.loc[df_sales['Product'] == 'Monitor']

if not monitor_rows.empty:
    val3_4 = monitor_rows.iloc[0]['Sales']
    print("\n3.4 First Sales value where Product = Monitor:")
    print(val3_4)
else:
    val3_4 = None
    print("\nNo Monitor records found.")
```

```python
# Task 4: Data Cleaning (Handling Missing Values)
print("\n--- Task 4: Data Cleaning (Handling Missing Values) ---")

# Use 'df_sales' from Task 2

# 4.1 Check missing values in each column
missing_counts4_1 = df_sales.isna().sum()

print("\n4.1 Missing values per column:")
print(missing_counts4_1)


# 4.2 Fill missing Sales values with the mean of Sales column
df4_2_filled = df_sales.copy()

sales_mean = df4_2_filled['Sales'].mean()

df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(sales_mean)

print("\n4.2 Missing values after filling Sales:")
print(df4_2_filled['Sales'].isna().sum())


# 4.3 Drop rows that contain any missing values
df4_3_dropped = df_sales.dropna()

print("\n4.3 Shape after dropping rows with missing values:")
print(df4_3_dropped.shape)


# Optional verification
print("\nRemaining missing values after dropna():")
print(df4_3_dropped.isna().sum())
```

```python
# Task 5: Data Manipulation (Columns, Sorting, Grouping)
print("\n--- Task 5: Data Manipulation (Columns, Sorting, Grouping) ---")

# Use 'df_sales' (the original one with NaNs)

# 5.1 Create a new column 'Total_Revenue'
# NaN values automatically propagate

df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

print("\n5.1 DataFrame with Total_Revenue:")
print(df_sales[['Sales', 'Quantity', 'Total_Revenue']].head())


# 5.2 Sort by Date (ascending) and then Sales (descending)

df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)

print("\n5.2 Sorted DataFrame:")
print(df5_2_sorted.head())


# 5.3 Group by Region and calculate sum of Sales and Quantity

df5_3_grouped_region = df_sales.groupby('Region')[
    ['Sales', 'Quantity']
].sum()

print("\n5.3 Total Sales and Quantity by Region:")
print(df5_3_grouped_region)


# 5.4 Group by Product and find average Sales

df5_4_grouped_product = df_sales.groupby('Product')[
    'Sales'
].mean()

print("\n5.4 Average Sales by Product:")
print(df5_4_grouped_product)
```
