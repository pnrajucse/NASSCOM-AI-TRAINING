```python
# Student exam score vectors: [Math, Science, English]

student1 = [85, 90, 88]
student2 = [78, 82, 80]
student3 = [92, 95, 91]

print("Student 1 Scores:", student1)
print("Student 2 Scores:", student2)
print("Student 3 Scores:", student3)
```

```python
import numpy as np

# Student score vectors
student1 = np.array([85, 90, 88])
student2 = np.array([78, 82, 80])
student3 = np.array([92, 95, 91])

# L2 Norms (Overall Performance)
norm1 = np.linalg.norm(student1)
norm2 = np.linalg.norm(student2)
norm3 = np.linalg.norm(student3)

print("L2 Norms:")
print("Student 1:", norm1)
print("Student 2:", norm2)
print("Student 3:", norm3)

# Function to compute cosine similarity
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Pairwise cosine similarities
sim12 = cosine_similarity(student1, student2)
sim13 = cosine_similarity(student1, student3)
sim23 = cosine_similarity(student2, student3)

print("\nCosine Similarities:")
print("Student1 vs Student2:", sim12)
print("Student1 vs Student3:", sim13)
print("Student2 vs Student3:", sim23)
```
