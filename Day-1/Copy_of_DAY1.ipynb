```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
# We'll imagine this is sensor data from various devices.
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0), # Celsius
    'Humidity_Sensor': (30.0, 80.0),   # Percentage
    'Pressure_Sensor': (980.0, 1030.0),# hPa
    'Light_Sensor': (100, 1000)        # Lux
}

# --- Task 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries, Indexing & Slicing ---
print("\n--- Task 1: Data Generation and Basic Structures ---")

# Generate 100 simulated sensor readings

raw_sensor_data = []

for i in range(1, 101):
    # Generate timestamp
    timestamp = datetime.datetime.now() - datetime.timedelta(
        minutes=random.randint(0, 10000)
    )
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    # Generate unique device ID
    device_id = f"DEV{i:03d}"

    # Randomly select device type
    device_type = random.choice(DEVICE_TYPES)

    # Randomly select location
    location = random.choice(LOCATIONS)

    # Generate sensor value based on device type range
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    # Create tuple and add to list
    reading = (timestamp_str, device_id, device_type, location, value)
    raw_sensor_data.append(reading)

# Display first 5 records
print("\nFirst 5 Sensor Readings:")
for record in raw_sensor_data[:5]:
    print(record)

print(f"\nTotal Sensor Readings Generated: {len(raw_sensor_data)}")
```

```python
# Create a list of dictionaries containing only temperature sensor readings

temperature_readings = [
    {
        'timestamp': reading[0],
        'device_id': reading[1],
        'location': reading[3],
        'temperature_celsius': reading[4]
    }
    for reading in raw_sensor_data
    if reading[2] == 'Temperature_Sensor'
]

# Display first 5 temperature readings
print("\nFirst 5 Temperature Readings:")
for temp in temperature_readings[:5]:
    print(temp)

print(f"\nTotal Temperature Readings: {len(temperature_readings)}")
```

```python
# Create a set of all unique locations

unique_locations = {reading[3] for reading in raw_sensor_data}

# Display the unique locations
print("\nUnique Locations:")
print(unique_locations)

print(f"\nTotal Unique Locations: {len(unique_locations)}")
```

```python
# Create a dictionary to store (min_value, max_value) for each device type

device_stats = {
    device_type: (float('inf'), float('-inf'))
    for device_type in DEVICE_TYPES
}

# Update min and max values based on sensor readings
for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device_type]

    device_stats[device_type] = (
        min(current_min, value),
        max(current_max, value)
    )

# Display the results
print("\nDevice Statistics (Min, Max):")
for device_type, stats in device_stats.items():
    print(f"{device_type}: {stats}")
```

```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")

# --- Global Data ---
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

# ==========================================================
# Task 1.1
# ==========================================================
raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(100):
    timestamp = (start_time + datetime.timedelta(minutes=i)).strftime(
        '%Y-%m-%d %H:%M:%S'
    )

    device_id = f"DEV{i+1:03d}"

    device_type = random.choice(DEVICE_TYPES)

    location = random.choice(LOCATIONS)

    min_val, max_val = READING_RANGES[device_type]

    value = round(random.uniform(min_val, max_val), 2)

    raw_sensor_data.append(
        (timestamp, device_id, device_type, location, value)
    )

# ==========================================================
# Task 1.2
# ==========================================================
temperature_readings = []

for reading in raw_sensor_data:
    if reading[2] == 'Temperature_Sensor':
        temperature_readings.append({
            'timestamp': reading[0],
            'device_id': reading[1],
            'location': reading[3],
            'temperature_celsius': reading[4]
        })

# ==========================================================
# Task 1.3
# ==========================================================
unique_locations = {reading[3] for reading in raw_sensor_data}

# ==========================================================
# Task 1.4
# ==========================================================
device_stats = {}

for device in DEVICE_TYPES:
    device_stats[device] = (float('inf'), float('-inf'))

for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device_type]

    device_stats[device_type] = (
        min(current_min, value),
        max(current_max, value)
    )

# ==========================================================
# Task 1.5.1
# ==========================================================
sliced_readings_1 = raw_sensor_data[9:20]
sliced_readings_2 = raw_sensor_data[::5]

# ==========================================================
# Task 1.5.2
# ==========================================================
sample_string = "PythonProgrammingIsFun"

substring_1 = sample_string[6:17]
substring_2 = sample_string[5::-1]

# ==========================================================
# Task 1.5.3
# ==========================================================
fifth_reading_device_type = raw_sensor_data[4][2]

device_type_suffix = fifth_reading_device_type[-6:]

# ==========================================================
# Task 2.1
# ==========================================================
def get_readings_by_location(readings_list, target_location):
    return [
        reading
        for reading in readings_list
        if reading[3] == target_location
    ]

# ==========================================================
# Task 2.2
# ==========================================================
def calculate_average_temperature(temp_data_list):
    if not temp_data_list:
        return 0.0

    total = sum(
        reading['temperature_celsius']
        for reading in temp_data_list
    )

    return total / len(temp_data_list)

# ==========================================================
# Task 2.3
# ==========================================================
def generate_report_line(reading_dict):
    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )

# ==========================================================
# Task 3.1
# ==========================================================
reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"Directory '{reports_dir}' created.")
else:
    print(f"Directory '{reports_dir}' already exists.")

# ==========================================================
# Task 3.2
# ==========================================================
if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    duration = last_dt - first_dt

    days = duration.days
    total_seconds = duration.total_seconds()

    hours = int(total_seconds // 3600) % 24
    minutes = int((total_seconds % 3600) // 60)

    print(
        f"Duration of data collection: "
        f"{days} days, {hours} hours, {minutes} minutes."
    )

# ==========================================================
# Task 3.3
# ==========================================================
random_values = [random.randint(100, 200) for _ in range(5)]

print("Random Values:", random_values)

# ==========================================================
# Task 4.1
# ==========================================================
report_filename = os.path.join(
    reports_dir,
    'temperature_report.txt'
)

try:
    with open(report_filename, 'w') as f:
        for reading_dict in temperature_readings:
            f.write(generate_report_line(reading_dict) + '\n')

    print(f"Temperature report written to '{report_filename}'.")

except IOError as e:
    print(f"Error writing file: {e}")

# ==========================================================
# Task 4.2
# ==========================================================
read_report_lines = []

try:
    with open(report_filename, 'r') as f:
        read_report_lines = f.readlines()

    print("\nFirst 3 lines of report:")

    for line in read_report_lines[:3]:
        print(line.strip())

except IOError as e:
    print(f"Error reading file: {e}")

# ==========================================================
# Sample Outputs
# ==========================================================
print("\nUnique Locations:", unique_locations)

print("\nDevice Statistics:")
for device, stats in device_stats.items():
    print(device, stats)

avg_temp = calculate_average_temperature(temperature_readings)
print(f"\nAverage Temperature: {avg_temp:.2f} °C")

print("\nExercise Complete!")
```
